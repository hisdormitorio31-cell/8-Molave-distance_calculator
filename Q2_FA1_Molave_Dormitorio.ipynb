```python
# 1. Define the functionss
def add_numbers(num1, num2):
    return num1 + num2

def subtract_numbers(num1, num2):
    return num1 - num2

def multiply_numbers(num1, num2):
    return num1 * num2

def divide_numbers(num1, num2):
    if num2 == 0:
        return "Cannot divide by zero."
    return num1 / num2

# 2. Ask the user for two numbers
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# 3. Ask the user to choose an operation
print("1 - Addition")
print("2 - Subtraction")
print("3 - Multiplication")
print("4 - Division")

choice = input("Enter choice: ")

# 4. Call the correct function and print the result
if choice == "1":
    result = add_numbers(num1, num2)
    print("Result:", result)
elif choice == "2":
    result = subtract_numbers(num1, num2)
    print("Result:", result)
elif choice == "3":
    result = multiply_numbers(num1, num2)
    print("Result:", result)
elif choice == "4":
    result = divide_numbers(num1, num2)
    print("Result:", result)
else:
    print("Invalid")

# Reflection

 # 1. What functions did you create in your calculater program?
 # - I created four functions: add_numbers, subtract_numbers, multiply_numbers, and divide_numbers.
 # 2. What parameters did your function use?
 # - Every function used two parameters: num1 and num2. These served as variables inside the functions to represent the numbers being calculated.
 # 3. What arguments were passed when the functions were called?
 # - When the functions were inside the if/elif blocks, the variables num1 and num2 were passed as arguments.
 # 4. How did your program use the returned value?
 # - The program took the value returned by the function and assigned it to the result variable. Then it used print to display the final output to the user,
 # 5. Why is it better to divide the calculator program into functions instead of writing everything in one long block of code?
 # - Dividing a program into functions improves readability, reusability, and maintainability by organizing logic into clean units instead of one long block of code.

```
